```python
import pandas as pd

df = pd.read_csv("../data/raw/Fraudulent_E-Commerce_Transaction_Data.csv")

df = df[df["Customer Age"] >= 0]

df.shape
```

```python
df["is_night"] = df["Transaction Hour"].apply(lambda x: 1 if (x < 6 or x >= 22) else 0)

df["is_night"].value_counts()
```

```python
df.groupby("is_night")["Is Fraudulent"].mean() * 100
```

```python
df["Account Age Days"].describe()
```

```python
df["is_new_account"] = df["Account Age Days"].apply(lambda x: 1 if x < 90 else 0)

df["is_new_account"].value_counts()
```

```python
df.groupby("is_new_account")["Is Fraudulent"].mean() * 100
```

```python
user_avg = df.groupby("Customer ID")["Transaction Amount"].transform("mean")

user_avg.head()
```

```python
user_avg = df.groupby("Customer ID")["Transaction Amount"].transform("mean")

user_avg.head()
```

```python
category_avg = df.groupby("Product Category")["Transaction Amount"].transform("mean")

category_avg.head()
```

```python
df["amount_vs_category_avg"] = df["Transaction Amount"] / category_avg

df["amount_vs_category_avg"].describe()
```

```python
df.groupby("Is Fraudulent")["amount_vs_category_avg"].mean()
```
